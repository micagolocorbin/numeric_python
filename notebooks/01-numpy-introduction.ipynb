## Overview

```python
# Numpy provides basic data structures - multidimensional arrays and matrices.

import numpy as np

# one dimensional Numpy array
temp = np.array([20.8, 21.9, 22.5, 22.7, 22.3, 21.0, 21.2, 20.9])
print(temp)
# convert temperature Celsius to degrees Fahrenheit
# simple scalar operations
degree = temp * 9 / 5 + 32
print(degree)

# This introduces one of the most important NumPy concepts: vectorization. Instead of explicitly looping over each element in Python, Numpy express the operation over the whole array.
# NumPy still processes the elements internally. Vectorization doesn't mean that no iteration occurs. It means that the iteration is handled by optimized NumPy operations rather than an explicit Python loop.
```

```python
# Implementation in Python
temp_list = [20.8, 21.9, 22.5, 22.7, 22.3, 21.0, 21.2, 20.9]
print(temp_list)
degree_list = [t * 9 / 5 + 32 for t in temp_list]
print(degree_list)
```

```python
# Numpy arrays are ndarrays or more precisely an instance of the class numpy.ndarray
print(type(temp))
print(type(temp_list))
```

## Comparison of NumPy Data Structures and Lists

```python
from sys import getsizeof as size
empty_lst = []
lst = [24,12,57,42]
size_of_list_object = size(lst)
size_of_element1 = size(lst[0])
size_of_element2 = size(lst[1])
size_of_element3 = size(lst[2])
size_of_element4 = size(lst[3])
total_list_size = size_of_list_object + size_of_element1 + size_of_element2 + size_of_element3 + size_of_element4

print("Memory usage of an empty list: ", size(empty_lst))
print("Size without size of the elements: ", size_of_list_object)
print("Size of all elements: ", size_of_element1 + size_of_element2 + size_of_element3 + size_of_element4)
print("Total size of the list: ", total_list_size)

# The memory requirement of a Python list consists of the size of the general list information, the memory required for the references to the list elements, and the size of all elements in the list.
```

```python
empty_arr = np.array([])
arr = np.array([24,12,57,42])
print("Memory requirement for the empty ndarray: ", size(empty_arr))
print("Total size of the ndarray: ", size(arr))
# When we define a NumPy array, NumPy automatically chooses a fixed integer size, in our case “int64”.
arr8 = np.array([24,12,57,42], np.int8)
arr16 = np.array([24,12,57,42], np.int16)
arr32 = np.array([24,12,57,42], np.int32)
arr64 = np.array([24,12,57,42], np.int64)
print(size(arr8), size(arr16), size(arr32), size(arr64))

# A Python list generally stores references to Python objects, while a standard numerical NumPy array stores values in a compact, homogeneous data buffer.
```

```python
# dtype (short for data type): How the bytes in memory should be interpreted by an array. 
print(arr.dtype)

# size: The total number of elements in the array across all dimensions.
print(arr.size)

# itemsize: The memory size of one individual element in bytes
print(arr.itemsize)
print(arr8.itemsize)

# nbytes: The total memory consumed by all elements in the array; nbytes = size x itemsize
print(arr.nbytes)

# shape: The array shape is a tuple of integers that represents the number of elements along each axis (dimension). 
print(arr.shape)
```

## Time Comparison Between Lists and NumPy Arrays

```python
import time

size_of_vec = 1_000_000
py_x = range(size_of_vec)
py_y = range(size_of_vec)
np_x = np.arange(size_of_vec)
np_y = np.arange(size_of_vec)

def pure_python():
  t1 = time.time()
  [x + y for x, y in zip(py_x, py_y)]
  return time.time() - t1

def pure_numpy():
  t1 = time.time()
  np_x + np_y
  return time.time() - t1

py_time = pure_python()
np_time = pure_numpy()
print(py_time)
print(np_time)
print(f"Numpy is {py_time / np_time:5.2f} times faster in this case.")
# :5.2f ensures the resulting float has a total width of at least 5 characters (including the decimal point) and is rounded to 2 decimal places.
```

```python
from timeit import Timer

size_of_vec = 1_000_000
py_X = range(size_of_vec)
py_Y = range(size_of_vec)
np_X = np.arange(size_of_vec)
np_Y = np.arange(size_of_vec)

def python_version():
  return [x + y for x, y in zip(py_X, py_Y)]

def numpy_version():
  return np_X + np_Y

# Set up the timers
timer_obj1 = Timer("python_version()", "from __main__ import python_version")
timer_obj2 = Timer("numpy_version()", "from __main__ import numpy_version")

# Run each benchmark 10 times and store the results
py_time = timer_obj1.timeit(10)
np_time = timer_obj2.timeit(10)

print(f"Python time (10 runs): {py_time:.5f} seconds")
print(f"NumPy time  (10 runs): {np_time:.5f} seconds")
print(f"NumPy is {py_time / np_time:.2f} times faster at size {size_of_vec}.")
```
